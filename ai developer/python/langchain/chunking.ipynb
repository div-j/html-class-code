```python
from langchain_ollama import ChatOllama
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python

text = """The Eiffel Tower, located in Paris, France, was constructed between 1887 and 1889 as the entrance arch to the 1889 World’s Fair.
It was designed by Gustave Eiffel’s engineering company and initially faced criticism from artists and intellectuals who considered it an eyesore.
Today, the Eiffel Tower is one of the most visited monuments in the world, attracting over 7 million visitors annually.
In 2015, special lighting systems were added to enhance its nighttime appearance and improve energy efficiency."""
```

```python

# 1. Sentence / recursive splitting/chunking
splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=20,
    separators=["\n\n", "\n", ". ", " ", ""]
)
chunks = splitter.split_text(text)

print("=== Chunks ===")
for i, c in enumerate(chunks, 1):
    print(f"{i}. {c}\n")
```

```python
import re

# Source Document
document = """The Eiffel Tower, located in Paris, France, was constructed between 1887 and 1889 as the entrance arch to the 1889 World’s Fair.
It was designed by Gustave Eiffel’s engineering company and initially faced criticism from artists and intellectuals who considered it an eyesore.
Today, the Eiffel Tower is one of the most visited monuments in the world, attracting over 7 million visitors annually.
In 2015, special lighting systems were added to enhance its nighttime appearance and improve energy efficiency."""

# --- STRATEGY 1: Sentence Splitting ---
def sentence_chunking(text):
    # Split text by periods followed by spaces or newlines
    sentences = re.split(r'\.\s+|\.\n', text.strip())
    # Re-add period to each sentence
    return [s.strip() + "." if not s.endswith(".") else s.strip() for s in sentences if s]

# --- STRATEGY 2: Semantic Chunking ---
def semantic_chunking(sentences):
    # Group by theme: History (Sentences 1-2) vs. Modern Day (Sentences 3-4)
    chunk_1_history = " ".join(sentences[0:2])
    chunk_2_modern = " ".join(sentences[2:4])
    return {
        "Chunk 1: Historical Context & Construction": chunk_1_history,
        "Chunk 2: Modern Popularity & Upgrades": chunk_2_modern
    }

# Execute Chunking
sentence_chunks = sentence_chunking(document)
semantic_chunks = semantic_chunking(sentence_chunks)

print("=== 1. SENTENCE SPLITTING RESULTS ===")
for i, chunk in enumerate(sentence_chunks, 1):
    print(f"Sentence Chunk {i}: {chunk}")

print("\n=== 2. SEMANTIC CHUNKING RESULTS ===")
for title, text in semantic_chunks.items():
    print(f"\n[{title}]\n{text}")
```
