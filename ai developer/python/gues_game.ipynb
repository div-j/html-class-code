### Guess the Number game

```python
import random
```

```python
def guess_the_number():
    print("Welcome to the Customizable Guess the Number Game!")
    
    # Allow player to choose custom range
    low = int(input("Enter minimum number (e.g., 1): "))
    high = int(input("Enter maximum number (e.g., 100): "))
    max_attempts = int(input("Enter max attempts allowed: "))
    
    secret_number = random.randint(low, high)
    attempts = 0
    
    print(f"\nI'm thinking of a number between {low} and {high}. You have {max_attempts} attempts!")
    
    while attempts < max_attempts:
        guess = int(input(f"Attempt {attempts + 1}/{max_attempts} - Enter your guess: "))
        attempts += 1
        
        if guess > secret_number:
            print("Your guess is too high.")
        elif guess < secret_number:
            print("Your guess is too low.")
        else:
            print(f"\nCongratulations! You guessed the number correctly in {attempts} attempt(s)!")
            return
            
    print(f"\nGame Over! You ran out of attempts. The secret number was {secret_number}.")
```

```python
if __name__ == "__main__":
    guess_the_number()
```
