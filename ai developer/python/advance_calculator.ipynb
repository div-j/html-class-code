```python
import math

# Basic Operation Functions
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        raise ZeroDivisionError("Division by zero is not allowed.")
    return a / b

# Advanced Operation Functions
def power(a, b):
    return math.pow(a, b)

def square_root(a, b=None):
    if a < 0:
        raise ValueError("Cannot compute the square root of a negative number.")
    return math.sqrt(a)

def logarithm(a, b=math.e):
    if a <= 0 or (b is not None and b <= 0):
        raise ValueError("Logarithm domain error: Inputs must be positive numbers.")
    if b is not None and b == 1:
        raise ValueError("Logarithm base cannot be 1.")
    return math.log(a) if b is None else math.log(a, b)


class Calculator:
    def __init__(self):
        # Initialize dictionary with basic operations
        self.operations = {
            '+': add,
            '-': subtract,
            '*': multiply,
            '/': divide
        }

    def add_operation(self, symbol, func):
        """Adds a new operation symbol and its corresponding function to the dictionary."""
        self.operations[symbol] = func

    def calculate(self, num1, symbol, num2=None):
        """Calculates result based on inputs with input validation and error handling."""
        # Validate that inputs are numbers using isinstance()
        if not isinstance(num1, (int, float)):
            print("Error: First input must be a valid number.")
            raise TypeError("First input value is not a number.")

        if num2 is not None and not isinstance(num2, (int, float)):
            print("Error: Second input must be a valid number.")
            raise TypeError("Second input value is not a number.")

        # Validate if operation symbol exists in the dictionary
        if symbol not in self.operations:
            print(f"Error: Invalid operation symbol '{symbol}'.")
            raise KeyError(f"Operation '{symbol}' is not supported.")

        # Execute calculation using the dictionary
        try:
            operation_func = self.operations[symbol]
            if num2 is None:
                return operation_func(num1)
            return operation_func(num1, num2)
        except Exception as e:
            print(f"Calculation Error: {e}")
            raise e


# Main Program
def main():
    calc = Calculator()

    # Register advanced operations using add_operation method
    calc.add_operation('^', power)
    calc.add_operation('sqrt', square_root)
    calc.add_operation('log', logarithm)

    print("Welcome to the OOP Calculator!")
    print("Available operations:", ", ".join(calc.operations.keys()))

    while True:
        symbol = input("\nEnter operation symbol (or 'exit' to quit): ").strip()
        if symbol.lower() == 'exit':
            print("Goodbye!")
            break

        try:
            num1_input = input("Enter first number: ")
            num1 = float(num1_input)

            num2 = None
            # Square root requires only 1 number
            if symbol != 'sqrt':
                num2_input = input("Enter second number (or press Enter for log default base e): ")
                if num2_input.strip() != "":
                    num2 = float(num2_input)

            # Perform calculation
            result = calc.calculate(num1, symbol, num2)
            print(f"Result: {result}")

        except ValueError:
            print("Error: Invalid numeric input provided.")
        except Exception as e:
            # Errors handled inside calc.calculate() are caught here to keep loop running
            continue


if __name__ == "__main__":
    main()
```
