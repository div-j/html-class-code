```python
# Question 1: Multiply all items in a list
sample_list = [2, 3, 6]
result = 1
for item in sample_list:
    result *= item
print("Q1:", result)

# Question 2: Sort list of tuples by last element
sample_tuples = [(2, 5), (1, 2), (4, 4), (2, 3), (2, 1)]
sorted_tuples = sorted(sample_tuples, key=lambda x: x[-1])
print("Q2:", sorted_tuples)

# Question 3: Combine dictionaries by summing common keys
d1 = {'a': 100, 'b': 200, 'c': 300}
d2 = {'a': 300, 'b': 200, 'd': 400}
combined = d1.copy()
for key, value in d2.items():
    combined[key] = combined.get(key, 0) + value
print("Q3:", combined)

# Question 4: Generate dictionary of squares from 1 to n
n = 8
square_dict = {i: i * i for i in range(1, n + 1)}
print("Q4:", square_dict)

# Question 5: Sort a tuple by float element
items = [('item1', '12.20'), ('item2', '15.10'), ('item3', '24.5')]
sorted_items = sorted(items, key=lambda x: float(x[1]), reverse=True)
print("Q5:", sorted_items)

# Question 6: Create a set, iterate, add, and remove items
my_set = {0, 1, 2, 3, 4}
print("Q6a:", my_set)

print("Q6b:")
for item in my_set:
    print(item)

my_set.add(5)
my_set.remove(0)
print("Q6c:", my_set)
```
