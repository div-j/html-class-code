```python
import faiss
import numpy as np

from sentence_transformers import SentenceTransformer
```

```python
corpus = [
    "Artificial Intelligence is transforming industries by automating tasks and improving decision making.",
    
    "Machine learning allows computers to learn patterns from data and make predictions.",
    
    "FAISS is a library designed for efficient similarity search over dense vectors.",
    
    "Python is widely used for data science, machine learning, and artificial intelligence.",
    
    "Artificial Intelligence can help manufacturers predict equipment failures before they happen.",
    
    "Businesses use machine learning to understand customer behavior and improve their products.",
    
    "Vector embeddings represent text as numerical vectors that capture semantic meaning.",
    
    "Deep learning uses neural networks to solve complex problems involving images, text, and speech.",
    
    "Data quality is important because machine learning models depend heavily on the data used for training.",
    
    "AI-powered systems are being used in healthcare, finance, education, manufacturing, and transportation."
]
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")

embeddings = model.encode(
    corpus,
    convert_to_numpy=True
)
```

```python
print(embeddings.shape)
```

```python
d = embeddings.shape[1]
index = faiss.IndexFlatL2(d)
```

```python
index.add(np.array(embeddings))
print("Number of vectors:", index.ntotal)
```

```python
query = "How can I use AI in industry?"
query_vector = model.encode(
    [query],
    convert_to_numpy=True
)
```

```python
D, I = index.search(
    query_vector,
    k=2
)

print("Top Results:")

for idx in I[0]:
    print(corpus[idx])
```
