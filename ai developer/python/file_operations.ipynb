```python
import numpy as np

# Filepath to your dataset
file_path = "Loan_prediction_dataset.csv"

# 1. Open the file using the open() function
file_stream = open(file_path, "r")

try:
    # 2. Use genfromtxt to load numerical data using comma delimiter
    data = np.genfromtxt(file_stream, delimiter=",", skip_header=1, filling_values=0)

    loan_amounts = data[:, 8]

    # 3. Calculate mean, median, and standard deviation
    mean_val = np.mean(loan_amounts)
    median_val = np.median(loan_amounts)
    std_val = np.std(loan_amounts)

    # Display statistical output
    print("--- Loan Amount Statistics ---")
    print(f"Mean Loan Amount: {mean_val:.2f}")
    print(f"Median Loan Amount: {median_val:.2f}")
    print(f"Standard Deviation: {std_val:.2f}")

finally:
    # 4. Explicitly close the file after reading
    file_stream.close()
    print("\nFile stream closed successfully.")
```
