### Langgraph

```python
from langgraph.graph import StateGraph
from langchain_ollama import ChatOllama
```

```python
def ask_question(state):
    """
    Function to ask a question to the Ollama model and return the response.
    """
    # Initialize the Ollama chat model
    chat_model = ChatOllama( model="llama3.2")
    
    # Get the response from the model
    response = chat_model.invoke("Give me a motivational quote")
    
    # Return the content of the response
    return {"output": response.content}
```

```python
graph = StateGraph(dict)
graph.add_node("start", ask_question)
graph.set_entry_point("start")
graph.set_finish_point("start")
```

```python
app = graph.compile()
result = app.invoke({})
print(result)  # This will print the output from the Ollama model
```
